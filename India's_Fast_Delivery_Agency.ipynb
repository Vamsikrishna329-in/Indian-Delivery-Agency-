```python
print(df.isnull().sum())
df = df.fillna(0)
```

```python
delivery_time_filtered_df = df[df['Delivery Time (min)'] > 30]
print(delivery_time_filtered_df)
```

```python
discount_applied = df[df['Discount Applied'] == 'Yes']
print(discount_applied)
```

### Bar Graph: Number of Reviews per Delivery Agent

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Count the number of reviews for each agent
agent_review_counts = df['Agent Name'].value_counts().reset_index()
agent_review_counts.columns = ['Agent Name', 'Number of Reviews']

# Create the bar graph for review counts
plt.figure(figsize=(12, 7))
sns.barplot(x='Agent Name', y='Number of Reviews', hue='Agent Name', data=agent_review_counts, palette='viridis', legend=False)
plt.title('Number of Reviews per Delivery Agent')
plt.xlabel('Delivery Agent')
plt.ylabel('Number of Reviews')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Bar Graph: Average Rating per Delivery Agent

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the average rating for each agent
agent_average_ratings = df.groupby('Agent Name')['Rating'].mean().reset_index()
agent_average_ratings.columns = ['Agent Name', 'Average Rating']

# Create the bar graph for average ratings
plt.figure(figsize=(12, 7))
sns.barplot(x='Agent Name', y='Average Rating', hue='Agent Name', data=agent_average_ratings, palette='magma', legend=False)
plt.title('Average Rating per Delivery Agent')
plt.xlabel('Delivery Agent')
plt.ylabel('Average Rating')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Bar Graph: Number of Orders by Order Type

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Count the number of orders for each order type
order_type_counts = df['Order Type'].value_counts().reset_index()
order_type_counts.columns = ['Order Type', 'Number of Orders']

# Create the bar graph for order type counts
plt.figure(figsize=(12, 7))
sns.barplot(x='Order Type', y='Number of Orders', hue='Order Type', data=order_type_counts, palette='Paired', legend=False)
plt.title('Number of Orders by Order Type')
plt.xlabel('Order Type')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Bar Graph: Number of Orders by City

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Count the number of orders for each city
city_order_counts = df['Location'].value_counts().reset_index()
city_order_counts.columns = ['City', 'Number of Orders']

# Create the bar graph for city order counts
plt.figure(figsize=(12, 7))
sns.barplot(x='City', y='Number of Orders', hue='City', data=city_order_counts, palette='viridis', legend=False)
plt.title('Number of Orders by City')
plt.xlabel('City')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Bar Graph: Average Delivery Time by Order Type

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the average delivery time for each order type
average_delivery_time_by_order_type = df.groupby('Order Type')['Delivery Time (min)'].mean().reset_index()
average_delivery_time_by_order_type.columns = ['Order Type', 'Average Delivery Time (min)']

# Create the bar graph for average delivery time by order type
plt.figure(figsize=(12, 7))
sns.barplot(x='Order Type', y='Average Delivery Time (min)', hue='Order Type', data=average_delivery_time_by_order_type, palette='coolwarm', legend=False)
plt.title('Average Delivery Time by Order Type')
plt.xlabel('Order Type')
plt.ylabel('Average Delivery Time (min)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Bar Graph: Average Delivery Time by City

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the average delivery time for each city
average_delivery_time_by_city = df.groupby('Location')['Delivery Time (min)'].mean().reset_index()
average_delivery_time_by_city.columns = ['City', 'Average Delivery Time (min)']

# Create the bar graph for average delivery time by city
plt.figure(figsize=(12, 7))
sns.barplot(x='City', y='Average Delivery Time (min)', hue='City', data=average_delivery_time_by_city, palette='coolwarm', legend=False)
plt.title('Average Delivery Time by City')
plt.xlabel('City')
plt.ylabel('Average Delivery Time (min)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Box Plot: Distribution of Delivery Times by City

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 8))
sns.boxplot(x='Location', y='Delivery Time (min)', data=df, palette='viridis', hue='Location', legend=False)
plt.title('Distribution of Delivery Times by City')
plt.xlabel('City')
plt.ylabel('Delivery Time (min)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
